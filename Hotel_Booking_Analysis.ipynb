# **Executive Summary**

Python, Pandas, and Seaborn to explore booking patterns, cancellation behavior, pricing trends, and customer segments. The goal is to identify actionable insights that can help hotels improve booking management, reduce cancellations, and support data-driven business decisions.

# ***Key Finding***

**hotel bookings were canceled.**
1. **Hotel Comparison:** City Hotel recorded higher cancellation rate than Resort Hotel.

2.**Booking Volume:** City Hotel received approximately twice as many bookings as Resort Hotel.

3.**Monthly Bookings:** August recorded the highest number of bookings (13,861), while January recorded the lowest (5,921) across the available data.

4.**Seasonality:** Booking volumes varied across months, but the partial coverage of 2015 and 2017 limits definitive seasonal conclusions.

# **Business Recommendations**

1. **Reduce Booking Cancellations**

With an overall cancellation rate of 37.08%, hotels should improve booking confirmation procedures and introduce pre-arrival reminders to help reduce cancellations.

2. **Focus on City Hotel Cancellations**

Since City Hotel has a higher cancellation rate than Resort Hotel, management should investigate the main reasons for cancellations and develop targeted retention strategies.

3. **Improve Staffing and Resource Planning**

Booking volumes vary significantly across months. Hotels should use demand forecasts to plan staffing, housekeeping, and operational resources more efficiently.

4. **Optimize Pricing Strategies**

Hotels should evaluate ADR differences across hotel types, booking lead times, and lengths of stay to identify opportunities for more effective pricing strategies.

5. **Monitor High-Risk Customer Segments**

Management should regularly analyze cancellation rates across market segments and booking channels to identify areas where booking policies or customer communication could be improved.
